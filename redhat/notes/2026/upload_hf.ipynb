```python
from huggingface_hub import HfApi

api = HfApi()

# 1. 配置你的参数
local_file_path = "/content/qwen3-0.6b-q4.gguf" # 你在 Colab 里的路径
repo_id = "wangzheng422/qwen3-0.6b-emoji-GGUF"  # 你的目标仓库名
filename_in_repo = "qwen3-0.6b-q4.gguf"        # 仓库中的文件名

# 2. 创建仓库（如果仓库不存在的话）
try:
    api.create_repo(repo_id=repo_id, repo_type="model", exist_ok=True)
    print(f"✅ 仓库 {repo_id} 已准备就绪")
except Exception as e:
    print(f"ℹ️ 仓库创建提示: {e}")

# 3. 开始上传
print(f"🚀 正在上传 {local_file_path} ... 请稍候")
api.upload_file(
    path_or_fileobj=local_file_path,
    path_in_repo=filename_in_repo,
    repo_id=repo_id,
    repo_type="model",
)

print(f"🎉 上传成功！你的模型地址: https://huggingface.co/{repo_id}/blob/main/{filename_in_repo}")
```
